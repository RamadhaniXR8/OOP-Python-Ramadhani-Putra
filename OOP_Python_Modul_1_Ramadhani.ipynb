```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name # Nama Hero
        self.hp = hp # Nyawa (Health Point)
        self.attack_power = attack_power # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# -- Main Program --

# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()
```

HP hero1 berubah menjadi 500 karena atribut hp masih bisa diubah langsung.

```python
# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")

hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

hero2 dikirim sebagai objek agar hero1 bisa mengakses atribut dan method milik hero2.

```python
# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")
```

super() digunakan untuk memanggil constructor dari class induk Hero.

```python
# -- Main Program Baru --

print("\n--- Update Class Hero ---")

eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama

        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0 # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# -- Uji Coba --

hero1 = Hero("Layla", 100)

# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)

hero1.set_hp(-50) # Coba set negatif
print(hero1.get_hp()) # Output: 0 (Karena dicegat oleh logika Setter)
```

HP menjadi 0 karena setter mencegah HP bernilai negatif.

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# 2. Implementasi pada Class Konkret

class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --

# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
h.serang("Serigala")
m.serang("Alucard")
```

Akan error karena GameUnit adalah abstract class yang tidak bisa dibuat menjadi objek langsung.

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --

# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]


print("--- PERANG DIMULAI ---")

# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```

Program tetap bisa menggunakan satu perintah serang(), tetapi hasilnya berbeda sesuai jenis hero.

```python
from abc import ABC, abstractmethod


# ABSTRACT CLASS / PARENT CLASS

class KamarHotel(ABC):

    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = stok
        self.__harga_dasar = harga_dasar

    # Getter untuk melihat stok
    def get_stok(self):
        return self.__stok

    # Getter untuk melihat harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Setter / Method untuk menambah stok
    def tambah_stok(self, jumlah):
        if self.__stok + jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! "
                f"Stok tidak boleh negatif ({self.__stok + jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok {self.nama_kamar}: "
                f"{self.__stok} unit."
            )

    # Abstract Method
    @abstractmethod
    def tampilkan_detail(self):
        pass

    # Abstract Method
    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# CLASS KAMAR DELUXE

class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    # Override method tampilkan_detail()
    def tampilkan_detail(self):
        print(
            f"[DELUXE] {self.nama_kamar} | "
            f"Fasilitas: {self.fasilitas}"
        )

    # Override method hitung_harga_total()
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.10
        harga_per_malam = harga_dasar + pajak
        total = harga_per_malam * jumlah_malam

        print(
            f"Harga Dasar/Malam: Rp {harga_dasar:,.0f} | "
            f"Pajak(10%): Rp {pajak:,.0f}"
        )
        print(
            f"Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {total:,.0f}"
        )

        return total


# CLASS KAMAR STANDARD

class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    # Override method tampilkan_detail()
    def tampilkan_detail(self):
        print(
            f"[STANDARD] {self.nama_kamar} | "
            f"Kapasitas: {self.kapasitas}"
        )

    # Override method hitung_harga_total()
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.05
        harga_per_malam = harga_dasar + pajak
        total = harga_per_malam * jumlah_malam

        print(
            f"Harga Dasar/Malam: Rp {harga_dasar:,.0f} | "
            f"Pajak(5%): Rp {pajak:,.0f}"
        )
        print(
            f"Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {total:,.0f}"
        )

        return total


# FUNGSI PROSES TRANSAKSI

def proses_transaksi(daftar_pesanan):
    total_tagihan = 0

    print("--- STRUK PEMESANAN ---")

    for nomor, pesanan in enumerate(daftar_pesanan, start=1):
        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["jumlah_malam"]

        print(f"{nomor}. ", end="")
        kamar.tampilkan_detail()

        subtotal = kamar.hitung_harga_total(jumlah_malam)

        total_tagihan += subtotal

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("----------------------------------------")


# MAIN PROGRAM

print("--- SETUP DATA KAMAR ---")

# Admin membuat 1 Kamar Deluxe
kamar_deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1500000,
    "Private Pool"
)

# Admin membuat 1 Kamar Standard
kamar_standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500000,
    "2 Orang"
)


# Admin menambahkan stok Kamar Deluxe
kamar_deluxe.tambah_stok(10)

# Admin mencoba menambahkan stok negatif
kamar_standard.tambah_stok(-5)

# Admin menambahkan stok Kamar Standard
kamar_standard.tambah_stok(20)

# PEMESANAN

daftar_pesanan = [
    {
        "kamar": kamar_deluxe,
        "jumlah_malam": 2
    },
    {
        "kamar": kamar_standard,
        "jumlah_malam": 1
    }
]


# Proses transaksi
proses_transaksi(daftar_pesanan)
```
